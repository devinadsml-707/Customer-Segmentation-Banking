
### Objective: Predicting cluster of a credit card customer at Berlian Bank!


1. First we need to import the libraries

```python
# Import library
import pickle
import json
import dill
import pandas as pd
import numpy as np
```

2. Then we load all the necessary pipeline, functions, and columns

```python
# Load model
with open('../models/pipelines.pkl', 'rb') as file_1:
    pipelines = pickle.load(file_1)

with open('../models/num_columns.txt', 'r') as file_2:
    num_columns = json.load(file_2)
```

3. After that, we create the inference data into a dictionary 

```python
# create data inference
data_inf = {
    'Customer ID': 9999, 
    'Balance': 2000, 
    'Balance Frequency': 0.5, 
    'Purchases': 1000,
    'One-off Purchases': 600, 
    'Installments Purchases': 800, 
    'Cash Advance': 500,
    'Purchases Frequency': 5, 
    'One-off Purchases Frequency': 0.5,
    'Purchases Installments Frequency': 0.7, 
    'Cash Advance Frequency': 0.3,
    'Cash Advance Transaction': 3, 
    'Purchases Transaction': 20, 
    'Credit Limit': 5000,
    'Payments': 800, 
    'Minimum Payments': 500, 
    'Full Payment Percentage': 0.2, 
    'Tenure': 12 
            }

data_inf = pd.DataFrame([data_inf])
data_inf
```

5. Check the pipeline for data inference

```python
pipelines
```

6. Predicts the data inference and cluster.

```python
print("="*100)
print("CLUSTER PREDICTION")
print("="*100)

#predict
y_pred_inf = pipelines['KMeans'].predict(data_inf[num_columns])[0]  # [0] to get single value

#map cluster number to label
cluster_mapping = {0: 'A', 1: 'B', 2: 'C', 3: 'D', 4: 'E'}
y_pred_label = cluster_mapping[y_pred_inf]

print(f"| Cluster Prediction: {y_pred_label} |")
```
